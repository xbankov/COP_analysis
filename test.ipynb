```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from src.utils.logger import setup_logger

logger = setup_logger()

# Configuration variables
headless = True
driver_path = "/home/xbankov/COP_analysis/chrome/chromedriver-linux64/chromedriver"
binary_path = "/home/xbankov/COP_analysis/chrome/chrome-linux64/chrome"

# Set up Chrome options
chrome_options = Options()
if binary_path:
    chrome_options.binary_location = binary_path  # Set the custom binary path
if headless:
    chrome_options.add_argument("--headless")  # Run Chrome in headless mode (no GUI)
    chrome_options.add_argument("--window-size=1920x1080")

# Set up the Chrome driver service
chrome_service = Service(driver_path)

# Initialize the Chrome browser
driver = webdriver.Chrome(service=chrome_service, options=chrome_options)
driver.implicitly_wait(20)

# Logging
logger.info("Chrome driver initialized with parameters:")
logger.info(f"Driver path: {driver_path}")
logger.info(f"Binary path: {binary_path}")
logger.info(f"Headless: {headless}")
logger.info(f"Default window-size: 1920x1080")

# Your Selenium code here...

url = "https://worldhealthorg.shinyapps.io/glass-dashboard/_w_bdea497f/_w_b3fce5de/#!/amr"
driver.get(url)

page_content = driver.page_source
logger.info("Page content retrieved successfully.")
print(page_content)

# Quit the driver after use
driver.quit()
```
